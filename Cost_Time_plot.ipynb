```python
import matplotlib.pyplot as plt

# Time values
time = [0, 1, 2, 3, 4, 5, 6]

# Cost values
cost = [0, 100, 180, 300, 450, 600, 800]

# Plot
plt.plot(time, cost, marker='o')

# Labels and title
plt.xlabel("Time")
plt.ylabel("Cost")
plt.title("Cost vs Time")

# Grid
plt.grid(True)

# Display graph
plt.show()
```
